4. Matrix Operation

```python
f=torch.randint(size=(2,3),low=0,high=10)
g=torch.randint(size=(3,2),low=0,high=10)
print(f)
print(g)
```

```python
# matrix mul:
torch.matmul(f,g)
```

```python
# dot product between two vectors
vector1=torch.tensor([1,2])
vector2=torch.tensor([3,4])
torch.dot(vector1,vector2)
```

```python
# traspose
torch.transpose(f,0,1)
```

```python
# deteminant
h=torch.randint(size=(4,4),low=0,high=10,dtype=torch.float32)
torch.det(h)
```

```python
# inverse:
torch.inverse(h)
```

7. INPLACE OPERATIONS

```python
# SUPPOSE WE ARE WORKING ON A LARGE DATASET AND THERE IS A PROBLEM OF SPACE AND WHENEVER YOU ARE GOING TO PERFORM ANY OPERATION BETWEEN ANY TWO MATRIX
# THE RESULT WILL BE STORED INTO A NEW TENSOR
```

```python
m=torch.rand(2,3)
n=torch.rand(2,3)
```

```python
m+n
```

```python
m.add_(n)
```

```python
m
```

```python
n
```

```python
m.relu_()
```

```python
# note: if you want to inplace permanent changes then pick any function and apply _ after it
```

8. Copying a Tensor

```python
a=torch.rand(2,3)
a
```

```python
# in python
b=a
b # both both are at same location so if you make changes to a it will reflect on b too
```
