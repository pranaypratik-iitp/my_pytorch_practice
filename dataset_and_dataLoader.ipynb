**problem with this manual code**

    1. No standard interface for data
    2. No easy way to apply teansformation
    3. shuffling and sampling
    4. batch management and parallelization






```python
# to solve above problems pytorch have 2 classes : 1. dataset 2. dataloader
from sklearn.datsets import make_classification
import torch
```

```python
# step 1 : creating a synthetic classification dataset using sklearn
X,y=make_classification(
    n_samples=10,          # number of samples
    n_features=2,          # no of features
    n_informative=2,       # no of informative features
    n_redundant=0,         # no of redundant feature
    n_classes=2,           # no of classes
    random_state=42.       # for reproducibility
)
```

```python
X.shape
```

```python
y.shape
```

```python
X=torch.tesor(X,dtype=torch.float32)
y=torch.tesor(y,dtype=torch.float32)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):
  def __init__(self,features,labels):
    self.features=features
    self.labels=labels

  def __len__(self):

    return self.features.shape[0]

  def __getitem__(self, index):
    return self.features[index], self.labels[index]
```

```python
dataset=CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataset[2]
```

```python
dataloader=DataLoader(dataset,batch_size=2,shuffle=True)
```

```python
for batch_features,batch_labels in dataloader:
  print(batch_features)
  print(batch_labels)
  print('-'*50)
```
