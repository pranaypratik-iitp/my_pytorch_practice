Designing a neural network having 2 cols 1. cgpa 2. placement and activation function used is sigmoid

```python
import torch
```

```python
x=torch.tensor(6.7)  # input feature
y=torch.tensor(0.0)  # true label(binary)

w=torch.tensor(1.0)  # weights
b=torch.tensor(0.0)  # bias
```

```python
# binary cross entropy loss for scalar
def binary_cross_entropy_loss(prediction,target):
  epsilon=1e-8      # to prevent log(0)
  prediction=torch.clamp(prediction,epsilon,1-epsilon)
  return -(target*torch.log(prediction)+(1-target)*torch.log(1-prediction))
```

```python
# forward pass:
z=w*x+b # weighted sum(linear part)
y_pred=torch.sigmoid(z)


# compute binary loss:
loss=binary_cross_entropy_loss(y_pred,y)
```

```python
# derivatives:
dloss_dy_pred=(y_pred-y)/(y_pred*(1-y_pred))
```

```python
# using autograd
x=torch.tensor(6.7)
y=torch.tensor(0.0)
```

```python
w=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)
```

```python
w
```

```python
b
```

```python
z=w*x+b
z
```

```python
y_pred=torch.sigmoid(z)
y_pred
```

```python
loss=binary_cross_entropy_loss(y_pred,y)
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```

**CLEARING GRADIENTS**

In pytorch gradients are not going to clear automatically as it starts accumulating like if we got 4 as gradient and again if we are going to calculate the same gradient then instead of 4 it will add 4 to it and the new grad will be 8.

```python
x.grad.zero_()  # clearing gradient
```
