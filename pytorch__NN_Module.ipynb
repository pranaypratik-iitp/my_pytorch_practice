*   The torch.nn module in pytorch is a core library that provides a wide array of classe & functions designed to help developers build neural network efficiently and effective.




**Key components of torch.nn**



1. modules(layers):

    1. nn.module : the base class for all the neural network modules. your custom models and layers should subclass this class
    2. common layer : icludes layers like nn.linear (fully connected layers) nn.conv2d(convolution layer) nn.lstm(recurrent layer) and many others



2. Activation function:


*   functions like ReLU, nn.sigmoid, nn.tanhintroduce non-linearity to the model, allowing it to learn complex patterns



3. loss function: provides loss function such as nn.crossEntropyLoss, nn.MSEloss and nn.NLLLOSS
2.  container modules : nn.Sequential


4. Regulation and dropout : layers like nn.dropout and nn.batchNorm2d helps preventing overfitting and improves the model's ability to generalize to new data










```python
# create model class: 5 inputs features with one neuron giving one output
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
  def __init__(self,num_features):          # num_features defines how much feature you want
    super().__init__()
    self.linear=nn.Linear(num_features,1)  # num_features - no. of input features and 1: no of output
    self.sigmoid = nn.Sigmoid()             # activation function

  def forward(self,features):
    out=self.linear(features)         # calculating z here means wx+b
    out=self.sigmoid(out)

    return out
```

```python
# create dataset
features=torch.randn(10,5)      # dummy dataset with 10 rows and 5 columns (equals to num of fearures)

# create model
model=Model(features.shape[1])  # created object of the model and features.shape[1] - signifies the num of features

# call model for forward pass

model(features)
```

```python
import torch
import torch.nn as nn
class Model(nn.Module):
  def __init__(self,num_features):          # num_features defines how much feature you want
    super().__init__()
    self.linear1=nn.Linear(num_features,3)  # num_features - no. of input features and 3: no of output
    self.relu = nn.ReLU()             # activation function
    self.linear2=nn.Linear(3,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear1(features)         # calculating z here means wx+b
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid(out)

    return out
```

```python
# create dataset
features=torch.randn(10,5)      # dummy dataset with 10 rows and 5 columns (equals to num of fearures)

# create model
model=Model(features.shape[1])  # created object of the model and features.shape[1] - signifies the num of features

# call model for forward pass

model(features)
```

```python
import torch
import torch.nn as nn
class Model(nn.Module):
  def __init__(self,num_features):          # num_features defines how much feature you want
    super().__init__()
    self.network=nn.Sequential(
        nn.Linear(num_features,3),
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid()
    )

  def forward(self,features):
    out=self.network(features)         # calculating z here means wx+b


    return out
```

```python
# create dataset
features=torch.randn(10,5)      # dummy dataset with 10 rows and 5 columns (equals to num of fearures)

# create model
model=Model(features.shape[1])  # created object of the model and features.shape[1] - signifies the num of features

# call model for forward pass

model(features)
```
